```python
def remove_page_duplicates(data, target_column):
    def clean_journey(journey):
        if pd.isna(journey):
            return journey
        
        pages = journey.split('-')
        if not pages:
            return journey
        
        # Keep first page and any page different from previous
        cleaned = [pages[0]]
        for page in pages[1:]:
            if page != cleaned[-1]:
                cleaned.append(page)
        
        return '-'.join(cleaned)
    
    # Create a copy of the data
    cleaned_data = data.copy()
    
    # Apply cleaning function to the target column
    cleaned_data[target_column] = cleaned_data[target_column].apply(clean_journey)
    return cleaned_data
```

```python
def group_by(data, group_column='user_id', target_column='user_journey',sessions='All', count_from='last'):
    df = data.copy()
    if 'session_id' in df.columns:
        df = df.sort_values([group_column, 'session_id'])
    else:
        # If no session_id, just sort by group_column
        df = df.sort_values(group_column)
        
    if sessions != 'All' and isinstance(sessions, int):
        # Filter to first N or last N sessions per user
        
        if count_from == 'last':
            # Take last N sessions for each user
            df = df.groupby(group_column).tail(sessions)
        elif count_from == 'first':
            # Take first N sessions for each user
            df = df.groupby(group_column).head(sessions)
        else:
            raise ValueError("count_from must be 'first' or 'last'")
            
    grouped = df.groupby(group_column)[target_column].apply(lambda x: '-'.join(x.astype(str))).reset_index()
    return grouped




```
