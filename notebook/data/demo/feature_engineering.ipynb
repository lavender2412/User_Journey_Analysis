```python
# check for the user conversion, if they have a Check-out option in the list, it should taken as converted.
# this will be our y-variable and once the value is actually determined for the known data, i.e., if it is found in the user journey
# we just say it got converted and after than we perform feature engineering and understand what is happening after that
def label_conversions(data, target_column='user_journey', conversion_page='Checkout'):
    data_copy = data.copy()
    data_copy['converted'] = data_copy[target_column].str.contains(conversion_page, case=False,  # Case-insensitive matching
        na=False).astype(int)   # Treat NaN values as False).astype(int)
    return data_copy


# now generating a conversion summary of the same

def get_conversion_summary(data, converted_column = 'converted'):
    total_users = len(data)
    converted_users = data[converted_column].sum()
    not_converted_users = total_users - converted_users
    conversion_rate = (converted_users / total_users) *100 if total_users > 0 else 0

    summary = pd.DataFrame({
        'Metric' : ['Total Users', 'Converted Users', 'Not Converted', 'Conversion Rate (%)'],
        'Value' : [total_users, converted_users, not_converted_users, round(conversion_rate, 2)]
    })

    return summary
```
